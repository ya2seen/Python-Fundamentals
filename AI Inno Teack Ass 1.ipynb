## 1. Password Validator

```python
password = input("Enter a password: ")

has_digit = False
has_uppercase = False

for character in password:
    if character.isdigit():
        has_digit = True
    if character.isupper():
        has_uppercase = True

if len(password) >= 8 and has_digit and has_uppercase:
    print("Valid Password")
else:
    print("Invalid Password")
```

## 2. Count Characters in a String

```python
sentence = input("Enter a sentence: ")

vowels = 0
consonants = 0
digits = 0
spaces = 0

for character in sentence:
    if character.isalpha():
        if character.lower() in "aeiou":
            vowels += 1
        else:
            consonants += 1
    elif character.isdigit():
        digits += 1
    elif character == " ":
        spaces += 1

print("Vowels:", vowels)
print("Consonants:", consonants)
print("Digits:", digits)
print("Spaces:", spaces)
```

## 3. Student Grades List

```python
grades = [75, 48, 90, 62, 35, 88, 51]

passed = 0
failed = 0

for grade in grades:
    print(grade)
    if grade >= 50:
        passed += 1
    else:
        failed += 1

print("Passed students:", passed)
print("Failed students:", failed)
```

## 4. Find the Largest and Smallest Number

```python
numbers = [15, 42, 8, 91, 23, 4, 55]

largest = numbers[0]
smallest = numbers[0]

for number in numbers:
    if number > largest:
        largest = number
    if number < smallest:
        smallest = number

print("Largest:", largest)
print("Smallest:", smallest)
```

## 5. Word Frequency Dictionary

```python
sentence = input("Enter a sentence: ")

words = sentence.lower().split()
word_counts = {}

for word in words:
    if word in word_counts:
        word_counts[word] += 1
    else:
        word_counts[word] = 1

print(word_counts)
```

## 6. Remove Duplicate Values

```python
numbers = [5, 2, 5, 8, 2, 10, 8, 12]

unique_numbers = []

for number in numbers:
    if number not in unique_numbers:
        unique_numbers.append(number)

print(unique_numbers)
```

## 7. Group Words by First Letter

```python
words = ["apple", "banana", "avocado", "blueberry", "cherry", "apricot"]

grouped_words = {}

for word in words:
    first_letter = word[0]
    if first_letter not in grouped_words:
        grouped_words[first_letter] = []
    grouped_words[first_letter].append(word)

print(grouped_words)
```

## 8. Shopping Cart

```python
cart = {
    "Laptop": 25000,
    "Mouse": 500,
    "Keyboard": 1200,
    "Headphones": 900
}

total_price = 0
expensive_products = 0

for product, price in cart.items():
    print(product, ":", price)
    total_price += price
    if price > 1000:
        expensive_products += 1

print("Total price:", total_price)
print("Products costing more than 1000:", expensive_products)

if total_price > 25000:
    print("Expensive Order")
else:
    print("Normal Order")
```

## 9. Common Values Between Two Lists

```python
list1 = [1, 2, 3, 4, 5, 8]
list2 = [3, 4, 5, 6, 7, 8]

common_values = []

for value in list1:
    if value in list2 and value not in common_values:
        common_values.append(value)

print("Common list:", common_values)
print("Common tuple:", tuple(common_values))
```

## 10. Student Management Task

```python
students = {
    "Ahmed": [70, 80, 90],
    "Sara": [95, 88, 92],
    "Omar": [40, 55, 45],
    "Mona": [60, 75, 70]
}

passed_students = []
failed_students = []

for name, grades in students.items():
    total = 0
    for grade in grades:
        total += grade
    average = total / len(grades)

    if average >= 50:
        print(name, "- Average:", round(average, 2), "- Passed")
        passed_students.append(name)
    else:
        print(name, "- Average:", round(average, 2), "- Failed")
        failed_students.append(name)

print("Passed students:", passed_students)
print("Failed students:", failed_students)
```
